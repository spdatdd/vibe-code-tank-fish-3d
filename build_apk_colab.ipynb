# Build APK - Bể cá 3D (Three.js + Capacitor)
Chạy tuần tự các cell bên dưới hoặc chọn **Runtime -> Run all** để đóng gói `index.html`, `lib/`, `assets/` thành file `app-debug.apk`.

```python
# 1. Cài Java 21, Node.js 20 và Android SDK (API 34/35)
!apt-get update -qq
!apt-get install -y -qq openjdk-21-jdk unzip wget
!curl -fsSL https://deb.nodesource.com/setup_20.x | bash -
!apt-get install -y -qq nodejs

!mkdir -p /content/android-sdk/cmdline-tools
!wget -q https://dl.google.com/android/repository/commandlinetools-linux-11076708_latest.zip -O /tmp/cmdline.zip
!unzip -q -o /tmp/cmdline.zip -d /content/android-sdk/cmdline-tools
!rm -rf /content/android-sdk/cmdline-tools/latest
!mv /content/android-sdk/cmdline-tools/cmdline-tools /content/android-sdk/cmdline-tools/latest

import os
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-21-openjdk-amd64"
os.environ["ANDROID_HOME"] = "/content/android-sdk"
os.environ["PATH"] += ":/content/android-sdk/cmdline-tools/latest/bin:/content/android-sdk/platform-tools"

!yes | sdkmanager --licenses > /dev/null
!sdkmanager "platform-tools" "platforms;android-35" "build-tools;35.0.0"
```

```python
# 2. Clone source code mới nhất từ GitHub và gom vào thư mục www/
%cd /content
!rm -rf /content/app
!git clone https://github.com/spdatdd/vibe-code-tank-fish-3d.git app
%cd /content/app

!mkdir -p www
!cp index.html www/
!cp -r lib assets www/
!ls -la www/
```

```python
# 3. Khởi tạo Capacitor & thêm nền tảng Android
%cd /content/app
!npm install --silent @capacitor/core @capacitor/cli @capacitor/android
!yes n | npx cap init "Be Ca 3D" "com.spdatd.beca3d" --web-dir=www
!npx cap add android
!npx cap sync android
```

```python
# 4. Build APK bằng Gradle và tải về máy
%cd /content/app/android
!chmod +x gradlew
!./gradlew assembleDebug --no-daemon

from google.colab import files
files.download("/content/app/android/app/build/outputs/apk/debug/app-debug.apk")
```
